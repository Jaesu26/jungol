# Gold Random Defense

## 갱스

- 문제 출처: [정올 2534번](https://jungol.co.kr/problem/2534)

`-` 분리 집합을 사용해서 친구 관계를 관리하자. 정보가 없는 경우 개별 갱으로 취급하는 게 가능한 갱의 수를 늘리는 데 이득이다. 따라서 $M$개의 사실을 기반으로 유니온을 수행한 뒤의 고유한 집합의 개수가 가능한 갱의 수의 최댓값이다. 이때 고유한 집합의 개수는 각 노드의 루트 노드를 모은 집합의 크기와 같다 (루트가 다르면 서로 다른 집합임)

`-` 사실 관계를 기반으로 유니온 하는 방법을 정리하자. $F$ 쿼리라면 단순히 둘을 유니온 하면 된다. $E$ 쿼리라면 $p$의 적들과 $q$를 유니온하고 $q$의 적들과 $p$를 유니온하자. 그리고 서로를 적으로 추가하면 된다

`-` E번 쿼리가 입력됐을 때 원수 순회는 최악의 경우 $O(N)$이다. 따라서 전체 알고리즘의 시간 복잡도는 최악의 경우 $O\left(NM\right)$이다

```python
class DisjointSet:
    def __init__(self, n):
        self._parents = list(range(n))
        self._sizes = [1] * n

    def union(self, u, v):
        u_root = self.find(u)
        v_root = self.find(v)
        if u_root == v_root:
            return False
        if self._sizes[u_root] > self._sizes[v_root]:
            u_root, v_root = v_root, u_root
        self._parents[u_root] = v_root
        self._sizes[v_root] += self._sizes[u_root]
        return True

    def find(self, u):
        p_u = self._parents[u]
        if p_u != u:
            self._parents[u] = self.find(p_u)
        return self._parents[u]


def solution():
    N = int(input())
    M = int(input())
    man2enemies = [[] for _ in range(N)]
    dsu = DisjointSet(N)
    for _ in range(M):
        op, p, q = input().split()
        p, q = int(p) - 1, int(q) - 1
        if op == "F":
            dsu.union(p, q)
            continue
        for q_e in man2enemies[q]:
            dsu.union(p, q_e)
        for p_e in man2enemies[p]:
            dsu.union(q, p_e)
        man2enemies[p].append(q)
        man2enemies[q].append(p)
    s = set([dsu.find(u) for u in range(N)])
    answer = len(s)
    print(answer)


solution()
```
