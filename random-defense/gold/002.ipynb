# Gold Random Defense

## List of Unique Integers

- 문제 출처: [정올 3017번](https://jungol.co.kr/problem/3017)

### 정답 풀이 (큐, 해시를 사용한 집합과 맵)

`-` $n$번째 원소까지 고려하여 가능한 연속 부분 수열의 경우의 수를 계산했다고 해보자

`-` 이때 $n+1$번째 원소인 $a_{n+1}$을 추가로 고려했을 때 경우의 수가 어떻게 되는지 보자

`-` 일단 $a_{n+1}$을 연속 부분 수열에 포함시키지 않는다면 의미가 없으므로 $a_{n+1}$은 연속 부분 수열에 포함시켜야 한다

`-` 그럼 $a_{n+1}$로 인해 새롭게 고려되는 연속 부분 수열은 $\{a_{n+1}\}, \{a_n, a_{n+1}\} ,  \{a_{n-1}, a_n, a_{n+1}\} ,\dots, \{a_i, \dots, a_{n-1}, a_n, a_{n+1}\}$와 같다 (단, $a_{i-1} = a_{n+1}$이며 연속 부분 수열의 모든 원소는 고유하다)

`-` 최장 연속 부분 수열의 길이는 $n-i+2$이다. 따라서 새롭게 추가된 연속 부분 수열의 개수는 $n-i+2$이다

`-` $n$번째 원소를 마지막 원소로 가지는 최장 연속 부분 수열을 고려하자. 그리고 해당 연속 부분 수열에 존재하는 고유 원소 집합을 가지고 있자

`-` 이제 $n+1$번째 원소를 최장 연속 부분 수열에 추가하려고 시도할 것이다

`-` 만약, $a_{n+1}$이 고유 원소 집합에 존재하지 않는다면 $a_{n+1}$을 문제 없이 추가할 수 있고 최장 연속 부분 수열의 길이만큼 경우의 수가 누적된다

`-` 이제 그렇지 않다고 해보자. 그럼 최장 연속 부분 수열의 왼쪽부터 $a_{n+1}$과 같은 값을 제거할 때까지 popleft를 수행하자

`-` 그리고 $a_{n+1}$을 추가한 뒤 길이만큼 경우의 수를 누적하면 된다

`-` queue로 관리하면 popleft를 $O(1)$에 수행할 수 있으며 모든 원소는 연속 부분 수열에 $1$번만 포함되고 최대 $1$번만 제거된다

`-` 따라서 전체 알고리즘의 시간 복잡도는 $O(N)$이다

`-` $23$분 걸렸고 $2$번 만에 맞혔다

`-` 원소를 popleft하면 고유 원소 집합에서도 제거해줘야 하는데 빼먹어서 $1$번 틀렸다

```python
from collections import deque


def solution():
    N = int(input())
    array = list(map(int, input().split()))
    count = 0
    subsequence = deque([])
    unique = set()
    for a in array:
        if a not in unique:
            unique.add(a)
        else:
            while subsequence[0] != a:
                x = subsequence.popleft()
                unique.remove(x)
            subsequence.popleft()
        subsequence.append(a)
        count += len(subsequence)
    print(count)


solution()
```

## 트리의 지름

- 문제 출처: [정올 4947번](https://jungol.co.kr/problem/4947)

### 정답 풀이 (트리, 그래프 탐색)

`-` 웰노운 문제이다. 임의의 정점에서 DFS를 수행하여 가장 먼 노드 $u$를 찾자. 트리이므로 BFS가 아닌 DFS로도 최장 거리를 찾을 수 있다

`-` 이때 노드 $u$에서 가장 멀리 떨어진 노드 $v$와의 거리가 트리의 지름이 된다

`-` 노드 $v$도 DFS를 수행해서 찾으면 되며 DFS를 $2$번 수행하므로 전체 알고리즘의 시간 복잡도는 $O(n)$이다

`-` $12$분 걸렸고 $1$번에 맞혔다

```python
def find_diameter(tree):
    n = len(tree)
    none, root = 0, 1
    distances = [-1] * n
    dfs(tree, root, none, distances)
    farthest_node_from_root = find_argmax(distances)
    distances = [-1] * n
    dfs(tree, farthest_node_from_root, none, distances)
    diameter = max(distances)
    return diameter


def dfs(tree, node, parent, distances):
    distances[node] = distances[parent] + 1
    for child in tree[node]:
        if child == parent:
            continue
        dfs(tree, child, node, distances)


def find_argmax(array):
    max_ = max(array)
    for i, a in enumerate(array):
        if a == max_:
            return i


def solution():
    n = int(input())
    tree = [[] for _ in range(n + 1)]
    for _ in range(n - 1):
        a, b = map(int, input().split())
        tree[a].append(b)
        tree[b].append(a)
    answer = find_diameter(tree)
    print(answer)


solution()

# input
# 2
# 1 2
```

## 문자들(LETTERS)

- 문제 출처: [정올 3529번](https://jungol.co.kr/problem/3529)

### 정답 풀이 (백트래킹)

`-` 이동 규칙 중 $1$번은 자명하며 $2$번은 $3$번에 종속된다. 따라서 $3$번 규칙만 유의하면 된다

`-` 각 셀에는 대문자 알파벳만이 들어있다. 알파벳은 총 $26$개이니 최대로 얻을 수 있는 상품 개수도 $26$이다

`-` 여태까지 획득한 상품의 상태 공간을 집합으로 관리하며 셀을 이동하자. 이때 백트래킹을 사용해 $3$번 규칙에 따라 가지치기를 수행하면 탐색 시간을 대폭 줄일 수 있다

`-` 솔직히 최악의 경우에서 동작하는 알고리즘의 시간 복잡도는 모르겠다

`-` 참고로 왼쪽 상단은 시작 셀로 무조건 밟아야 함에 유의하자

`-` $21$분 걸렸고 $1$번에 맞혔다

```python
def backtracking(graph, r, c, n_rows, n_cols, drc, seen):
    global ANSWER
    n = len(seen)
    if n > ANSWER:
        ANSWER = n
    for dr, dc in drc:
        nr, nc = r + dr, c + dc
        is_in_range = 0 <= nr < n_rows and 0 <= nc < n_cols
        if not is_in_range:
            continue
        a = graph[nr][nc]
        if a in seen:
            continue
        seen.add(a)
        backtracking(graph, nr, nc, n_rows, n_cols, drc, seen)
        seen.remove(a)


def solution():
    global ANSWER
    R, C = map(int, input().split())
    board = [input() for _ in range(R)]
    drc = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    ANSWER = 0
    backtracking(board, 0, 0, R, C, drc, set(board[0][0]))
    print(ANSWER)


solution()
```

## 하노이6(초보들의 하노이)

- 문제 출처: [정올 3272번](https://jungol.co.kr/problem/3272)

### 정답 풀이 (동적 계획법)

`-` 각 원판이 어느 기둥에 위치하고 있는지와 현재 몇 번째 이동인지가 동일하면 현재 상태를 만드는 경우의 수 역시 동일하다

`-` 각 원판이 위치할 수 있는 기둥은 $3$개이므로 가능한 원판 배치의 수는 $O\left(3^N\right)$이다. $2$번 규칙에 의해 실제 경우의 수는 시간 복잡도에 비해 훨씬 작다

`-` 또, 각 기둥별로 가장 위의 있는 원판은 $O(N)$에 알 수 있으며 고유한 상태 공간을 만들기 위해 각 기둥별로 쌓여진 원판 번호를 정렬하여 가지고 있을 것이다. 이는 $O(N \log N)$이다

`-` 따라서 동적 계획법을 이용하면 $O\left(NM 3 ^ N \log N \right)$에 문제를 해결할 수 있다. 점화식을 전개할 때 $3$번 규칙에 유의하자

`-` $N$은 최대 $5$이고 $M$은 최대 $1000$이므로 제한 시간 안에 충분히 실행될 수 있다

`-` $43$분 걸렸고 $1$번에 맞혔다

```python
from collections import defaultdict


def hanoi(n, m, mod):
    dp = [defaultdict(int) for _ in range(m + 2)]
    dp[0][tuple(range(1, n + 1)), (), ()] = 1
    none = n + 1
    num_ways = 0
    for i in range(m + 1):
        for (first, second, third), count in dp[i].items():
            dp[i][first, second, third] %= mod
            if len(third) == n:
                num_ways += count
                num_ways %= mod
                continue
            min_first = min(first, default=none)
            min_second = min(second, default=none)
            min_third = min(third, default=none)
            state = {"first": first, "second": second, "third": third}
            if min_first < min_second:
                update(count, first, second, min_first, min_second, "first", "second", state.copy(), dp[i + 1])
            if min_first < min_third:
                update(count, first, third, min_first, min_third, "first", "third", state.copy(), dp[i + 1])
            if min_second < min_first:
                update(count, second, first, min_second, min_first, "second", "first", state.copy(), dp[i + 1])
            if min_second < min_third:
                update(count, second, third, min_second, min_third, "second", "third", state.copy(), dp[i + 1])
            if min_third < min_first:
                update(count, third, first, min_third, min_first, "third", "first", state.copy(), dp[i + 1])
            if min_third < min_second:
                update(count, third, second, min_third, min_second, "third", "second", state.copy(), dp[i + 1])
    return num_ways


def update(count, from_, to, min_from, min_to, from_type, to_type, state, dp):
    if min_from >= min_to:
        return
    _from, _to = list(from_), list(to)
    _from.remove(min_from)
    _to.append(min_from)
    _to.sort()
    state[from_type], state[to_type] = tuple(_from), tuple(_to)
    dp[tuple(state.values())] += count


def solution():
    N, M = map(int, input().split())
    mod = 10**9 + 7
    answer = hanoi(N, M, mod)
    print(answer)


solution()

# input
# 3 8
```

`-` 코드가 너무 더러워서 리팩토링 좀 했다

`-` $22$분 걸린 듯 (풀이 시간에 포함 안 시킴)

- 재미나이 피셜

`-` 원판을 옮길 때 기둥의 맨 위에 얹혀진다. 따라서 정렬은 $O(N\log N)$이 아니라 $O(N)$이다 (내부적으로 Timsort 사용, 맨 앞에 원소 하나 추가)

`-` 가능한 원판 배치의 수가 $O\left(\min\left(3^N, 3^M\right)\right)$이며 각 배치를 관리하는데 $O(N)$이 필요하므로 상태 공간의 크기는 엄밀하게 $O\left(N \cdot \min \left(3^N, 3^M\right)\right)$이다

## 팰린드롬 유형

- 문제 출처: [정올 11236번](https://jungol.co.kr/problem/11236)

### 정답 풀이 (두 포인터, 브루트 포스)

`-` 만약 $w$가 $\text{type-$0$}$ 팰린드롬이라면 $w$의 양 끝 문자는 동일하다

`-` 이에 착안해 $w$의 양 끝 문자를 보자. 만약 이 둘이 같다면 양 끝 문자를 없애도 $w$의 $\text{type-$k$}$는 변하지 않는다. 따라서 이 둘을 없앤 뒤 새로운 문자열에 대해 양 끝 문자를 재귀적으로 확인하자. 양 끝의 문자를 없애는 건 문자열의 양 끝을 가리키는 포인터를 반대편으로 $1$칸씩 옮기는 방식으로 구현할 수 있다

`-` 이제 $w$의 양 끝 문자가 다르다고 해보자. 그럼 $w$를 팰린드롬으로 만들기 위해 양 끝 문자 중 적어도 하나는 제거해야 한다. 양 끝 문자 중 하나를 제거한 문자열을 $w'$이라 하자

`-` $w'$이 $\text{type-$0$}$ 팰린드롬이라면 $w$는 $\text{type-$1$}$ 팰린드롬이다. 하지만 $w'$이 $\text{type-$0$}$ 팰린드롬이 아닐 수 있다

`-` 잘 생각해 보면 주어진 문자열이 $w$에서 $w'$으로 바뀌었을 뿐 나머지는 동일하다

`-` 따라서 동일한 논리에 따라 $w'$의 양 끝 문자를 확인해 나가면 된다

`-` 제거할 수 있는 문자 수는 최대 $3$이며 제거 방향은 둘 중 하나이다. 그러므로 가능한 경우의 수는 $2^3$이다

`-` 즉, 아무리 많아도 $8$개의 경우를 확인하면 $w$가 어떤 $\text{type-$k$}$ 팰린드롬인지 알 수 있다

`-` 따라서 전체 알고리즘의 시간 복잡도는 $O(n)$이다

`-` $25$분 걸렸고 $1$번에 맞혔다

```python
import sys

sys.setrecursionlimit(10**5 + 2)


def check_type_k(s, left, right, n_removals):
    if n_removals > 3:
        return INF
    if left > right:
        return 0
    if s[left] == s[right]:
        return check_type_k(s, left + 1, right - 1, n_removals)
    k1 = check_type_k(s, left + 1, right, n_removals + 1)
    k2 = check_type_k(s, left, right - 1, n_removals + 1)
    return 1 + min(k1, k2)


def solution():
    global INF
    INF = float("inf")
    w = input().rstrip()
    answer = check_type_k(w, 0, len(w) - 1, 0)
    if answer == INF:
        answer = -1
    print(answer)


solution()

# input
# abdeba
```

## 회의준비

- 문제 출처: [정올 1317번](https://jungol.co.kr/problem/1317)

### 정답 풀이 (너비 우선 탐색, 브루트 포스)

`-` 사람을 노드로 관계를 간선으로 나타내면 양방향 그래프가 된다

`-` $1$번과 $2$번 조건에 의해 위원회의 수는 컴포넌트의 수와 같다

`-` 각 위원회별로 모든 참석자들의 의사전달시간 중 최댓값을 최소로 해야 한다

`-` 컴포넌트 안에서 위원회의 대표를 시작으로 BFS를 수행하면 모든 참석자들의 의사전달시간을 알 수 있다

`-` 이때 컴포넌트 내 각 참석자들을 대표로 가정한 뒤 BFS를 수행하면 모든 참석자들의 의사전달시간 시간 중 최댓값을 최소로 하는 대표를 찾을 수 있다

`-` 이를 각 위원회별로 수행하면 문제의 정답을 구할 수 있다

`-` BFS의 시간 복잡도는 $O(N+M)$이며 이를 각 노드(사람)별로 수행해야 하므로 전체 알고리즘의 시간 복잡도는 $O\left(N^2 + NM\right)$이다

`-` $31$분 걸렸고 $1$번에 맞혔다

```python
from collections import deque


def minimize_communication_time(graph):
    n = len(graph)
    component_ids = [None] * n
    owners = {}
    min_distances = {}
    id_ = 0
    for u in range(1, n):
        visited = bfs(graph, u)
        d = max(visited.values())
        if component_ids[u] is None:
            for node in visited:
                component_ids[node] = id_
            id_ += 1
        i = component_ids[u]
        if d < min_distances.get(i, n):
            owners[i] = u
            min_distances[i] = d
    return owners


def bfs(graph, start):
    queue = deque([start])
    visited = {start: 0}
    while queue:
        u = queue.popleft()
        for v in graph[u]:
            if v in visited:
                continue
            visited[v] = visited[u] + 1
            queue.append(v)
    return visited


def solution():
    N = int(input())
    M = int(input())
    graph = [[] for _ in range(N + 1)]
    for _ in range(M):
        u, v = map(int, input().split())
        graph[u].append(v)
        graph[v].append(u)
    owners = minimize_communication_time(graph)
    print(len(owners))
    for owner in sorted(owners.values()):
        print(owner)


solution()

# input
# 1
# 0
```

## 조각난 님

- 문제 출처: [정올 12692번](https://jungol.co.kr/problem/12692)

### 정답 풀이 (게임 이론, 애드 혹)

`-` 일반적인 님 게임과 달리 어떤 더미에서 돌을 제거 할지를 상대방이 정해준다. 물론 제거할 개수는 내가 정한다

`-` $n$이 작은 경우에서 시뮬레이션을 돌려보며 깨달았는데 모든 더미에 돌이 $2$개 이상 있다면 무조건 선공이 이긴다

`-` 상대가 어떤 더미를 고르든 거기서 $1$개만 남기고 나머지를 제거하자. 그리고 상대방에겐 $1$개만 남은 더미를 골라주자. 그럼 상대방은 할 수 있는 선택이 하나뿐이다

`-` 이를 반복하면 결국엔 선공 차례에 더미가 하나만 남게 되며 이번엔 돌을 $1$개만 남기는 게 아니라 모든 돌을 가져가서 게임을 승리할 수 있다

`-` 위의 과정에서 알 수 있듯이 어떤 더미에 돌이 $1$개만 있다는 것은 선공과 후공을 바꾸는 것과 같다. 따라서 돌이 $1$개만 있는 더미의 수가 짝수라면 선공과 후공이 그대로이고 홀수라면 선공과 후공을 바꾸는 것이다

`-` 그러므로 돌이 $1$개만 있는 더미의 수가 짝수면 선공이 이기고 홀수면 후공이 이기게 된다

`-` 단, 모든 더미에 돌이 $1$개만 있다면 얘기가 조금 달라진다. 위에서 `돌을 1개만 남기는 게 아니라 모든 돌을 가져가서 게임을 승리할 수 있다`고 언급했었다. 그런데 모든 더미에 돌이 $1$개만 있다면 이것이 성립하지 않는다. 이미 $1$개만 남아있기 때문이다

`-` 따라서 이러한 경우엔 $n$이 짝수면 후공이 이기고 홀수면 선공이 이긴다 (이를 고려 안 해서 $1$번 틀렸음)

`-` $33$분 걸렸고 $2$번째 시도에 맞혔다

```python
def solve_testcase():
    n = int(input())
    array = list(map(int, input().split()))
    n_ones = len([a for a in array if a == 1])
    if n_ones == n:
        if n % 2 == 0:
            print("Bob")
        else:
            print("Alice")
        return
    if n_ones % 2 == 0:
        print("Alice")
        return
    print("Bob")


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 4
# 1 1 1 1
```

## 타일밟기

- 문제 출처: [정올 2024번](https://jungol.co.kr/problem/2024)

### 정답 풀이 (동적 계획법)

`-` 현재 $i$번째 타일에 위치하고 있다고 해보자. 공차와 여태까지 연속적으로 밟은 타일의 수가 같다면 앞으로 타일을 밟아서 얻는 합의 최댓값도 동일하다

`-` $\operatorname{dp}[i][k][d]$를 현재 $i$번째 타일에 위치하고 있고 공차가 $d$이며 현재 타일을 포함해 여태까지 $k$개의 타일을 밟았을 때 얻은 합의 최댓값이라 하자. 단, 타일은 연속적으로 $3$개만 밟을 수 있으면 되므로 $k=1,2,3$ 중 하나이다 ($4$ 이상은 무의미)

`-` 따라서 $\operatorname{dp}[i][\min(k + 1, 3)][d] = A_i + \max\limits_{1\le j < i}\{\operatorname{dp}[j][k][d]\}$, (단, $d = A_i - A_j)$. $A_i$는 $i$번째 타일에 적힌 자연수이다

`-` $k=1$이면 공차가 존재하지 않으니 예외 처리 해주면 된다. 초깃값으로 $k=2$일 때 가능한 경우를 모두 세팅해두자. 그럼 앞으로 공차가 존재하는 것만 고려할 수 있게 된다

`-` 현재 $i$번째 타일에 위치하고 있을 때 가능한 공차의 수는 $i-1$이다. 따라서 전체 알고리즘의 시간 복잡도는 $O\left(N^2\right)$이다

`-` 가능한 공차의 수는 $O(N)$이지만 가능한 값의 범위는 타일마다 상이하므로 배열이 아닌 딕셔너리로 관리하자

`-` $36$분 걸렸고 $1$번에 맞혔다

```python
def solution():
    N = int(input())
    tiles = list(map(int, input().split()))
    dp = [[{} for _ in range(4)] for _ in range(N)]
    for i in range(1, N):
        for j in range(i):
            d = tiles[i] - tiles[j]
            dp[i][2][d] = tiles[i] + tiles[j]
    for i in range(2, N):
        for j in range(1, i):
            d = tiles[i] - tiles[j]
            for k in range(2, 4):
                k_curr = min(k + 1, 3)
                if d not in dp[j][k]:
                    continue
                dp[i][k_curr][d] = max(tiles[i] + dp[j][k][d], dp[i][k_curr].get(d, 0))
    answer = max(max(dp[i][3].values(), default=0) for i in range(2, N))
    print(answer)


solution()

# input
# 3
# 1 2 3
```

## MooTube

- 문제 출처: [정올 4053번](https://jungol.co.kr/problem/4053)

### 정답 풀이 (트리에서의 동적 계획법)

`-` 각 노드를 루트로 가정한 뒤 동적 계획법을 사용해 다른 노드와의 연관도를 저장해두고 쿼리 때 사용하자

`-` 전체 알고리즘의 시간 복잡도는 $O\left(N^2 + NQ\right)$이다

`-` $N$과 $Q$는 최대 $5000$이고 제한 시간은 $8$초라 통과할 줄 알았는데 시간 초과가 발생했다

`-` 계속 고통 받다가 트리 탐색을 재귀가 아닌 반복문으로 구현하니 $6$번 시도 끝에 맞혔다...

`-` 총 $55$분 걸렸다

```python
def dfs(tree, root, k, dp):
    n = len(tree)
    stack = [(root, k)]
    visited = [False] * n
    visited[root] = True
    while stack:
        node, k = stack.pop()
        for child, r in tree[node]:
            if visited[child]:
                continue
            visited[child] = True
            k_c = k if k < r else r
            dp[root][child] = k_c
            stack.append((child, k_c))


def solution():
    N, Q = map(int, input().split())
    tree = [[] for _ in range(N + 1)]
    for _ in range(N - 1):
        p, q, r = map(int, input().split())
        tree[p].append((q, r))
        tree[q].append((p, r))
    inf = float("inf")
    none = 0
    dp = [[inf] * (N + 1) for _ in range(N + 1)]
    for root in range(1, N + 1):
        dfs(tree, root, inf, dp)
    answers = []
    for _ in range(Q):
        k, v = map(int, input().split())
        answer = sum(r >= k for u, r in enumerate(dp[v]) if u != none and u != v)
        answers.append(answer)
    print("\n".join(map(str, answers)))


solution()

# 2 1
# 1 2 3
# 1 2
```

## 맛있는 사과

- 문제 출처: [정올 12357번](https://jungol.co.kr/problem/12357)

### 정답 풀이 (오프라인 쿼리)

`-` 만약 쿼리가 큰 것부터 주어진다면 한번 조건을 만족하기 시작한 사과는 이후의 쿼리에서도 계속 조건을 만족한다

`-` 그럼 쿼리가 바뀜에 따라 덜 맛있는 사과들만 추가로 고려해주면 된다

`-` 이는 쿼리와 사과를 정렬함으로써 수행할 수 있다

`-` 전체 알고리즘의 시간 복잡도는 $O(N\log N + Q\log Q)$이다

`-` $11$분 걸렸고 $1$번에 맞혔다

```python
def solution():
    N, Q = map(int, input().split())
    tastes = list(map(int, input().split()))
    sizes = list(map(int, input().split()))
    apples = [(t, s) for t, s in zip(tastes, sizes)]
    apples.sort()
    queries = [(int(input()), i) for i in range(Q)]
    queries.sort(reverse=True)
    max_size = 0
    count = 0
    answers = [0] * Q
    for p, i in queries:
        while apples and apples[-1][0] >= p:
            _, s = apples.pop()
            if s == max_size:
                count += 1
            elif s > max_size:
                max_size = s
                count = 1
        answers[i] = count
    print("\n".join(map(str, answers)))


solution()

# input
# 1 1
# 3
# 3
# 3
```

## 철도역

- 문제 출처: [정올 3642번](https://jungol.co.kr/problem/3642)

### 정답 풀이 (이분 탐색)

`-` 철도역의 위치가 정해지면 버스 요금의 합을 $O(N)$에 계산할 수 있다

`-` 철도역의 위치는 이분 탐색으로 찾을 것이다. 비용 조건식에 따라 사람과 철도역 사이의 거리들이 비슷해야 이득이다 (같거나 적다)

`-` 철도역의 위치 $s$에 따른 버스 요금의 합을 함수 $f(s)$로 생각하면 이 함수는 하나의 최솟값을 가진다. 하지만 비용 조건식에 의해 이러한 최솟값을 가지는 위치가 연속하여 있을 수 있다. 즉, 하나의 구간이며 동일한 값을 가지는 구간은 최솟값 구간 외에 존재하지 않는다

`-` 따라서 $f(s) \le f(s - 1)$이고 $f(s) \le f(s+1)$이라면 $s$가 최적의 위치가 된다. 그렇지 않다면 $f(s - 1) < f(s) < f(s + 1)$ 이거나 $f(s + 1) < f(s) < f(s-1)$이 성립한다. 등호가 포함될 수 없는 것이 만약 그렇다면 $f(s) \le f(s - 1) \ge f(s + 1)$을 이미 만족하기 때문이다. 함수 $f$는 아래로 볼록한 함수임에 유의하자

`-` $31$분 걸렸고 $1$번에 맞혔다

```python
def binary_search(positions, costs):
    inf = float("inf")
    x_max = max(positions)
    left, right = 0, x_max
    while left <= right:
        mid = (left + right) // 2
        cost_mid = compute_total_cost(mid, positions, costs)
        cost_mid_m1 = compute_total_cost(mid - 1, positions, costs) if mid > 0 else inf
        cost_mid_p1 = compute_total_cost(mid + 1, positions, costs) if mid < x_max else inf
        if cost_mid <= cost_mid_m1 and cost_mid <= cost_mid_p1:
            return cost_mid
        if cost_mid < cost_mid_m1:
            left = mid + 1
        else:
            right = mid - 1


def compute_total_cost(s, positions, costs):
    total = 0
    for x in positions:
        total += costs[abs(s - x)]
    return total


def solution():
    L, N = map(int, input().split())
    C = [0] + list(map(int, input().split()))
    X = list(map(int, input().split()))
    answer = binary_search(X, C)
    print(answer)


solution()

# input
# 1 1
# 10
# 1
```

## 펭귄의 이글루 후보지

- 문제 출처: [정올 12227번](https://jungol.co.kr/problem/12227)

### 정답 풀이 (동적 계획법)

`-` 생각을 너무 오래했다. 최대 연속 부분합을 잘 활용하면 되는 문제이다

`-` $\operatorname{dp}[i]$를 $A_i$를 마지막 원소로 가지는 최대 연속 부분합이라 하자

`-` 그리고 $\operatorname{dp2}[i]$를 $A_i$를 첫 번째 원소로 가지는 최대 연속 부분합이라 하자

`-` 그럼 $i$번 발판을 포함하는 최대 연속 부분합은 $\operatorname{dp}[i] + \operatorname{dp2}[i] - A_i$가 되며 이것이 $K$ 이상이라면 이글루를 지을 수 있다

`-` $\operatorname{dp}$와 $\operatorname{dp2}$는 $O(N)$에 구성할 수 있으므로 전체 알고리즘의 시간 복잡도도 $O(N)$이다

`-` $31$분 걸렸고 $1$번에 맞혔다

```python
def solution():
    N, K = map(int, input().split())
    array = list(map(int, input().split()))
    dp = [0] * (N + 1)
    for i, a_i in enumerate(array, start=1):
        dp[i] = max(a_i, dp[i - 1] + a_i)
    dp2 = [0] * (N + 1)
    for i in range(N - 1, -1, -1):
        a_i = array[i]
        dp2[i] = max(a_i, dp2[i + 1] + a_i)
    for i, a_i in enumerate(array):
        print("YES" if dp[i + 1] + dp2[i] - a_i >= K else "NO")


solution()

# input
# 3 1
# 1 2 3
```

## 미술관람 대회

- 문제 출처: [정올 2765번](https://jungol.co.kr/problem/2765)

### 정답 풀이 (그래프 탐색)

`-` 기본적으로 그래프를 탐색하며 조건에 따라 방문할 수 있는 곳은 모두 방문한 뒤 방문했다고 표시할 것이다

`-` 각 좌표를 순회하며 아직 방문한 좌표가 아니라면 해당 지점에서 그래프 탐색을 수행하면 된다

`-` 그러면 그래프 탐색 함수를 호출한 횟수가 정답이 된다

`-` $22$분 걸렸고 $1$번에 맞혔다

`-` 급하게 한다고 변수명 오타를 너무 많이 내서 디버깅 하느라 오래 걸렸다...

```python
import sys
from itertools import product

sys.setrecursionlimit(10**4 + 2)


def is_same(c1, c2):
    return c1 == c2


def is_same_for_redgreen_colorblind(c1, c2):
    c = c1 + c2
    return c1 == c2 or c == "RG" or c == "GR"


def dfs(graph, r, c, drc, visited, color_condition):
    n = len(graph)
    visited.add((r, c))
    for dr, dc in drc:
        nr, nc = r + dr, c + dc
        is_in_range = 0 <= nr < n and 0 <= nc < n
        if not is_in_range:
            continue
        is_same_color = color_condition(graph[r][c], graph[nr][nc])
        if (nr, nc) in visited or not is_same_color:
            continue
        dfs(graph, nr, nc, drc, visited, color_condition)


def solution():
    N = int(input())
    graph = [input() for _ in range(N)]
    drc = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    visited = set()
    visited_blind = set()
    count = 0
    count_blind = 0
    for r, c in product(range(N), range(N)):
        if (r, c) not in visited:
            dfs(graph, r, c, drc, visited, is_same)
            count += 1
        if (r, c) not in visited_blind:
            dfs(graph, r, c, drc, visited_blind, is_same_for_redgreen_colorblind)
            count_blind += 1
    print(count, count_blind)


solution()

# input
# 2
# RR
# GG
```

## 계보

- 문제 출처: [정올 1181번](https://jungol.co.kr/problem/1181)

### 정답 풀이 (그리디, 재귀)

`-` 어떤 노드 $u$의 부모가 $x$명이라고 해보자

`-` 일단 $x \le d$이면 아무 문제가 없으니 $x > d$로 가정하자

`-` 각 선조마다 최대 $d$명씩 할당할 수 있으므로 $x=qd + r, (0\le r < d)$라 할 때 $q$명의 선조를 추가하면 $qd$명의 부모를 $q$명으로 줄일 수 있다. 그럼 남은 부모는 $q+r$명이 된다

`-` $q+r \le d$라면 문제 없지만 그렇지 않다면 선조가 추가로 필요하다

`-` 이는 처음 $x$에서 $q+r$이 됐을 뿐이므로 $x \le d$일 때까지 재귀적으로 반복하면 된다

`-` 즉, $f(x, d)$를 부모의 수를 $x$에서 $d$ 이하로 줄이기 위해 추가해야 하는 선조의 최소 수라고 하면 $f(x, d) = q + f(q + r, d)$이며 $x\le d$인 경우 $0$을 반환한다

`-` $q+r \approx \frac{x}{d}$이므로 전체 알고리즘의 시간 복잡도는 $O(n \log n)$이다

`-` $19$분 걸렸고 $1$번에 맞혔다

```python
def f(x, d):
    if x <= d:
        return 0
    q = x // d
    r = x % d
    return q + f(q + r, d)


def solution():
    n, d = map(int, input().split())
    children = list(map(int, input().split()))
    num_parents = [0] * (n + 1)
    for u in children:
        num_parents[u] += 1
    answer = sum(f(x, d) for x in num_parents)
    print(answer)


solution()

# input
# 4 2
# 0 0 0
```

## 정사각형

- 문제 출처: [정올 4856번](https://jungol.co.kr/problem/4856)

### 정답 풀이 (브루트 포스, 해시를 사용한 집합과 맵)

`-` 단순하게 네 점을 임의로 정한 뒤 이들이 정사각형을 이루는지 판단하는 건 $O\left(N^4\right)$이고 $N$이 최대 $3000$이므로 시간 초과이다

`-` 정사각형의 한 변을 구성하는 양 끝 점이 고정되면 정사각형의 특징(네 변의 길이가 같고 모든 각이 $90^{\circ}$)에 의해 나머지 두 점도 고정된다

`-` 모든 변이 $x$축 또는 $y$축과 평행한 정사각형을 생각한 뒤 한 점을 중심으로 정사각형을 회전시켜 보면 나머지 두 점의 좌표를 어떻게 찾을 수 있는지 알 수 있다

`-` 이는 $O(1)$에 계산할 수 있으며 점의 존재 여부도 해시를 사용하면 $O(1)$에 수행할 수 있다. 따라서 전체 알고리즘의 시간 복잡도는 $O\left(N^2\right)$이다

`-` $21$분 걸렸고 $1$번에 맞혔다

```python
from itertools import product


def solution():
    N = int(input())
    points = [tuple(map(int, input().split())) for _ in range(N)]
    point_set = set(points)
    max_area = 0
    for p1, p2 in product(points, points):
        x_gap = p2[0] - p1[0]
        y_gap = p2[1] - p1[1]
        p3 = (p1[0] + y_gap, p1[1] - x_gap)
        p4 = (p2[0] + y_gap, p2[1] - x_gap)
        if p3 not in point_set or p4 not in point_set:
            continue
        max_area = max(x_gap**2 + y_gap**2, max_area)
    print(max_area)


solution()

# input
# 4
# 1 1
# 1 2
# 2 1
# 2 2
```

- 제미나이 피드백

`1.` `product(points, points)` 대신 `itertools.combinations(points, 2)` 사용 $\longrightarrow$ **가독성**도 챙기고 **실행 시간**도 **절반**으로 단축된다

`2.` `x1, y1 = p1`과 같이 언패킹해서 사용하기

`3.` 변수명으로 `x_gap` 대신 `dx` 사용하기 (참고로 `dx`는 $\Delta x$를 의미한다)
