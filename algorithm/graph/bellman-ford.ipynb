# 벨만-포드 (Bellman-Ford)

`-` 음의 가중치가 존재하는 그래프에서 한 정점과 다른 정점들 사이의 최단 경로를 찾는 알고리즘

## 블랙홀

- 문제 출처: [정올 2223번](https://jungol.co.kr/problem/2223)

`-` 그래프에 음수 사이클이 존재하는지 판단하는 문제이다. 벨만-포드 알고리즘을 써서 해결할 수 있다

`-` 임의의 최단 경로는 최대 $V-1$개의 간선을 가진다. 따라서 $V - 1$번의 relaxation 이후에도 최단 거리가 줄어들면 음수 사이클이 존재하는 것이다

```python
def check_negative_cycle(graph, weights, source, n_nodes, max_weight):
    max_distance = max_weight * n_nodes
    distances = [max_distance] * (n_nodes + 1)
    distances[source] = 0
    for _ in range(n_nodes - 1):
        for u in range(1, n_nodes + 1):
            for v in graph[u]:
                w = weights[u, v]
                if distances[u] + w < distances[v]:
                    distances[v] = distances[u] + w
    for u in range(1, n_nodes + 1):
        for v in graph[u]:
            if distances[u] + weights[u, v] < distances[v]:
                return True
    return False


def solve_testcase(max_t):
    N, M, B = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    weights = {}
    for _ in range(M):
        S, E, T = map(int, input().split())
        graph[S].append(E)
        graph[E].append(S)
        weights[S, E] = min(T, weights.get((S, E), max_t))
        weights[E, S] = weights[S, E]
    for _ in range(B):
        S, E, T = map(int, input().split())
        graph[S].append(E)
        weights[S, E] = min(-T, weights.get((S, E), max_t))
    if check_negative_cycle(graph, weights, S, N, max_t):
        print("YES")
    else:
        print("NO")


def solution():
    max_t = 10000
    F = int(input())
    for _ in range(F):
        solve_testcase(max_t)


solution()

# input
# 1
# 3 2 1 
# 1 2 3 
# 2 3 4 
# 3 1 8
```

`-` 벨만-포드을 구현할 땐 굳이 인접 리스트를 사용하지 말고 모든 간선을 리스트 하나에 가지고 있자. 어차피 $V-1$번의 반복마다 모든 간선을 확인해야 한다
