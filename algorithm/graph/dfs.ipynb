# 깊이 우선 탐색 (Depth-First Search)

`-` 시작 노드에서 출발해 갈 수 있는 데까지 깊이 탐색한 뒤 막히면 이전 분기로 되돌아가서 다른 경로를 탐색하는 알고리즘

## 미로 탐색

- 문제 출처: [정올 1912번](https://jungol.co.kr/problem/1912)

`-` 임의의 노드 $u$와 연결된 노드를 방문하기 전 이들을 번호순으로 정렬하자

`-` 재귀 DFS를 사용해 간단하게 구현할 수 있다

```python
def dfs(graph, u, visited, order):
    visited.add(u)
    order.append(u)
    graph[u].sort()
    for v in graph[u]:
        if v not in visited:
            dfs(graph, v, visited, order)


def solution():
    N, M = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    for _ in range(M):
        u, v = map(int, input().split())
        graph[u].append(v)
        graph[v].append(u)
    source = 1
    visited = set()
    order = []
    dfs(graph, source, visited, order)
    print(*order)


solution()

# input
# 2 1
# 1 2
```
