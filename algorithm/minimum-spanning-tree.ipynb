# 최소 스패닝 트리 (Minimum Spanning Tree)

`-` 간선들의 가중치 합이 최소인 스패닝 트리. 무방향 그래프에서 모든 정점을 포함하는 트리를 스패닝 트리라고 한다

## 워터링 홀(Watering Hole)

- 문제 출처: [정올 5187번](https://jungol.co.kr/problem/5187)

`-` MST가 태깅되어 있는 단계별 문제이다

`-` 최소 한군데는 우물을 파야 된다. 한 곳을 정해서 우물을 팠다고 하자. 그럼 나머지에 대해 이곳과 연결할지 우물을 팔지 결정해야 한다. 이 중 비용이 가장 작은 것 하나만 선택한 뒤 작업을 반복하면 모든 목초지에 물을 공급하기 위해 지불해야 하는 최소 금액을 알 수 있다. 그런데 이는 총 $O\left(N^4\right)$이라 시간 초과이다

`-` 우물을 파든 이미 물이 있는 목초지와 연결을 하든 물을 공급하는 것은 같다는 것에 집중하자. 우물을 파는 건 처음에 우물을 판 곳과 연결하는 걸로 생각하자. 그럼 모든 목초지에 물을 공급하는 건 모두를 하나로 연결한 것과 동일한 의미를 가진다. 이때 최소 비용으로 연결해야 하므로 최소 스패닝 트리의 비용을 구하면 된다

`-` 초기 우물 공급으로 가능한 경우는 $N$개이다. 정렬된 간선들에 대해 크루스칼 알고리즘은 $O\left(N^2\right)$에 동작한다

`-` 정렬은 최초 $1$번만 수행하면 되므로 전체 알고리즘의 시간 복잡도는 $O\left(N^3\right)$이다

```python
def compute_mst_cost(sorted_edges, source, initial_cost, n_nodes):
    dsu = DisjointSet(n_nodes)
    union = dsu.union
    mst_cost = initial_cost
    for u, v, w in sorted_edges:
        if u == v == source:
            continue
        if u == v and union(u, source):
            mst_cost += w
            continue
        if union(u, v):
            mst_cost += w
    return mst_cost


def solution():
    N = int(input())
    self_costs = [int(input()) for _ in range(N)]
    connection_costs = [list(map(int, input().split())) for _ in range(N)]
    edges = []
    for u in range(1, N):
        for v in range(u):
            edges.append((u, v, connection_costs[u][v]))
    edges.extend([(u, u, w) for u, w in enumerate(self_costs)])
    edges.sort(key=lambda x: x[2])
    answer = float("inf")
    for source in range(N):
        mst_cost = compute_mst_cost(edges, source, self_costs[source], N)
        answer = min(mst_cost, answer)
    print(answer)


solution()

# input
# 2
# 10
# 2
# 0 5
# 5 0
```

`-` 더미 노드를 하나 만들어서 우물을 파는 걸 더미 노드와 union 하는 걸로 생각하면 초기 우물 공급을 신경쓰지 않아도 되므로 $O\left(N^2 \log N\right)$에 해결할 수 있다
