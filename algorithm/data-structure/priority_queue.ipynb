# 우선순위 큐 (Priority Queue)

`-` 원소가 들어온 순서에 관계없이 우선순위가 가장 높은 원소가 먼저 삭제되는 자료구조

`-` 대개 Heap으로 구현하며 파이썬에선 heapq 내장 모듈을 활용하면 된다. 힙의 크기가 $N$이고 원소 간 비교 비용이 $O(K)$일 때 삽입 및 우선순위가 가장 높은 원소의 삭제를 $O(K\log N)$에 수행할 수 있다

## 쇼핑몰

- 문제 출처: [정올 3337번](https://jungol.co.kr/problem/3337)

`-` 단계별 문제를 풀어보자. 대분류가 STL이다

`-` 각 계산대별 대기 종료 시점을 기록하자. 즉, (대기 종료 시점, 계산대 번호)를 원소로 가지고 있자. 그럼 가장 빠른 대기 종료 시점에 계산대 번호가 가장 낮은 계산대를 고객에게 안내하면 된다

`-` 또, 계산대별 계산 종료 시점을 기록하자. 즉, (계산 종료 시점, 계산대 번호, 고객 회원번호)를 원소로 가지고 있자. 가장 빠른 계산 종료 시점에 번호가 가장 큰 계산대의 고객이 먼저 쇼핑몰을 나간다

`-` 계산대별 대기 종료 시점은 처음에 $\{(0, 1), (0, 2), \dots, (0,k)\}$로 초기화되어 있으며 고객이 입장할 때마다 두 집합을 갱신하면 된다

`-` 집합을 우선순위 큐로 관리하자. 그럼 전체 알고리즘의 시간 복잡도는 $O(k + N \log (N + k))$이다

```python
import heapq


def solution():
    N, k = map(int, input().split())
    array = [list(map(int, input().split())) for _ in range(N)]
    in_pq, out_pq = [], []
    for i in range(1, k + 1):
        in_pq.append((0, i))
    heapq.heapify(in_pq)
    for id_i, w_i in array:
        t, j = heapq.heappop(in_pq)
        heapq.heappush(in_pq, (t + w_i, j))
        heapq.heappush(out_pq, (t + w_i, -j, id_i))
    answer = 0
    for i in range(1, N + 1):
        _, _, r_i = heapq.heappop(out_pq)
        answer += i * r_i
    print(answer)


solution()

# input
# 2 2
# 1 5
# 2 10
```

## 중앙값

- 문제 출처: [정올 1570번](https://jungol.co.kr/problem/1570)

`-` 웰노운 Running Median 문제이다. 매 쿼리마다 배열을 수정하는 건 $O\left(N^2 \right)$으로 비효율적이다. 대신 두 개의 힙을 사용하자

`-` 현재 중앙값이 $m$일 때 $m$ 이하의 수들을 관리하는 최대 힙과 $m$ 이상의 수들을 관리하는 최소 힙을 고려하자. 최대 힙은 원소 $i$개를 관리하고 최소 힙은 원소 $i-1$개를 관리하여 최대 힙의 top이 $m$이 되도록 할 것이다

`-` 새로운 두 수 $P,Q$가 들어오면 우선 $P$를 최대 힙에 넣자. 그리고 두 힙의 크기를 같게 하기 위해 최대 힙의 top을 최소 힙에 추가할 것이다. 이 과정에서 힙 간의 대소 관계도 정상화된다. 마지막으로 $Q$를 최소 힙에 넣은 뒤 최소 힙의 top을 최대 힙에 추가하면 된다. 그럼 최대 힙의 top이 중앙값을 가리키게 된다. 파이썬에서 제공하는 heapq 모듈은 최소 힙이므로 최대 힙을 구현할 땐 정수에 $-1$을 곱해서 사용해야 한다

`-` 힙에 원소를 추가하고 top을 삭제하는 건 $O(\log N)$이므로 전체 알고리즘의 시간 복잡도는 $O(N\log N)$이다

```python
import heapq


def solution():
    N = int(input())
    x = int(input())
    Q = (N - 1) // 2
    max_heap = [-x]
    min_heap = []
    answers = [x] * (Q + 1)
    for i in range(1, Q + 1):
        P, Q = map(int, input().split())
        heapq.heappush(max_heap, -P)
        heapq.heappush(min_heap, -heapq.heappop(max_heap))
        heapq.heappush(min_heap, Q)
        heapq.heappush(max_heap, -heapq.heappop(min_heap))
        answers[i] = -max_heap[0]
    print("\n".join(map(str, answers)))


solution()

# input
# 3
# 2
# 1 5
```

`-` 이 유형을 백준에서도 풀었지만 구현이 가장 깔끔해서 기록용으로 남긴다 (점점 발전됨)
