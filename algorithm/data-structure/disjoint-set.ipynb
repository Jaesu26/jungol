# 분리 집합 (Disjoint Set)

`-` 서로 겹치는 원소가 없는 여러 개의 집합을 관리하는 자료구조

## 비용

- 문제 출처: [정올 2467번](https://jungol.co.kr/problem/2467)

`-` DSU로 푸는 문제라고 한다. 모든 쌍에 대해 $\operatorname{Cost}$를 시뮬레이션으로 구하는 건 시간 초과이다

`-` $\operatorname{Cost}(u, v)$를 구하는 과정을 생각해보자. 간선을 계속 제거하다가 어떤 간선이 끊기면서 둘 사이의 경로가 사라지게 된다. 즉, 이전까진 $u$와 $v$가 같은 집합이었는데 끊긴 것이다. 이를 역으로 생각해보자

`-` 가중치가 가장 큰 간선이 마지막에 제거되면서 모든 경로가 사라진다. 따라서 역으로 생각하면 초기엔 아무 간선도 없었다. 여기서 시작해 가중치가 큰 순으로 간선을 추가해 나간다. 이 간선의 가중치가 $w$이고 $u$와 $v$를 잇는다면 $\operatorname{Cost}(u, v)$는 $\text{제거된 간선들의 가중치 총합} + w$이다. (단, $u < v$)

`-` 이를 확장해 생각하자. 연결되지 않은 $u,v$가 연결된다는 것은 $\operatorname{Cost}(u, v)$만 결정되는 게 아니다. 각 집합에 속한 노드간의 경로도 이를 기점으로 연결되므로 각 집합의 크기가 $su, sv$라면 가능한 $u,v$ 쌍은 총 $su \cdot sv$개이며 총 비용은 $su \cdot sv \cdot \operatorname{Cost}(u, v)$이 된다

`-` 따라서 역순으로 모든 간선이 추가될 때까지 순회하면 문제를 해결할 수 있다. 전체 알고리즘의 시간 복잡도는 $O(N + M \log M)$이다

```python
def solution():
    mod = 10**9
    N, M = map(int, input().split())
    edges = [list(map(int, input().split())) for _ in range(M)]
    edges.sort(key=lambda e: -e[2])
    dsu = DisjointSet(N)
    union, is_connected, get_size_of = dsu.union, dsu.is_connected, dsu.get_size_of
    remainder_weight = sum(map(lambda e: e[2], edges))
    total_cost = 0
    for x, y, w in edges:
        x, y = x - 1, y - 1
        remainder_weight -= w
        if is_connected(x, y):
            continue
        sx = get_size_of(x)
        sy = get_size_of(y)
        total_cost += sx * sy * (remainder_weight + w)
        total_cost %= mod
        union(x, y)
    print(total_cost)


solution()

# input
# 2 1
# 1 2 3
```

`-` 참고로 DSU 구현체는 다음 링크에 있는 걸 사용했다. 앞으로도 중복해서 사용할 것 같아서 코드에 명시적으로 포함하지 않겠다
