# 스택 (Stack)

`-` 원소를 추가하거나 가장 최근에 추가된 원소를 삭제하는 연산을 $O(1)$에 수행할 수 있는 후입선출 구조의 자료구조

`-` 파이썬에선 간편하게 리스트를 사용하면 된다 (보통 연결 리스트로 구현하나 파이썬의 리스트는 동적 배열로 구조가 다르다, 하지만 스택으로 사용하는 것에 무리는 없다)

## 가장 큰 수 만들기

- 문제 출처: [정올 3036번](https://jungol.co.kr/problem/3036)

`-` 앞자리부터 비교하므로 앞자리가 커야 된다

`-` 따라서 모노톤 스택을 활용해 숫자들을 내림차순으로 관리하자. $K > 0$인 동안은 삭제가 가능하다. 삭제 후 $K$에서 $1$을 차감하면 된다

`-` 순회를 끝낸 후 $K > 0$이라면 남은 $K$만큼 스택에서 원소를 삭제하자

`-` 전체 알고리즘의 시간 복잡도는 $O(N)$이다

```python
def solution():
    N, K = map(int, input().split())
    A = list(input().rstrip())
    stack = []
    for x in A:
        while stack and stack[-1] < x and K > 0:
            stack.pop()
            K -= 1
        stack.append(x)
    while K > 0:
        stack.pop()
        K -= 1
    answer = "".join(map(str, stack))
    print(answer)


solution()

# input
# 3 1
# 195
```
