# 누적 합 (Prefix Sum)

`-` 길이가 $N$인 배열에 대해 $O(N)$ 전처리 후 임의의 구간 합을 $O(1)$에 구하는 기법

## 누적 거리

- 문제 출처: [정올 4803번](https://jungol.co.kr/problem/4803)

`-` 단계별 문제를 풀어보자. 이 문제는 누적 합을 이용해서 해결할 수 있다

`-` 마을의 위치를 기준으로 오름차순 정렬하자. 어떤 $k$에 대해 $x_k \le x$이고 $x < x_{k + 1}$를 만족한다고 해보자. 그럼 $f(x)$를 다음과 같이 나타낼 수 있다

`-` $f(x) = a_1 \times (x - x_1) + \cdots + a_k \times (x - x_k) + \cdots + a_{k+1} \times (x_{k+1} - x) +\cdots + a_N \times (x_N - x) = x \sum\limits_{i=1}^{k}a_i - \sum\limits_{i=1}^{k}a_i x_i + \sum\limits_{i=k+1}^{N}a_i x_i - x \sum\limits_{i=k+1}^{N}a_i$

`-` $p_a(i) = \sum\limits_{j=1}^{i}a_j,\, p_{ax}(i) = \sum\limits_{j=1}^{i}a_j x_j$로 정의하면 $f(x) = x\cdot (2p_a(k) - p_a(N)) + p_{ax}(N) -  2p_{ax}(k)$이며 이는 $O(1)$에 계산할 수 있다. 올바른 $k$는 이분 탐색으로 $O(\log N)$에 찾을 수 있으니 쿼리당 $O(\log N)$이다

`-` 정렬은 $O(N \log N)$이고 누적 합 배열은 $O(N)$에 정의할 수 있으므로 전체 알고리즘의 시간 복잡도는 $O((N + Q)\log N)$이다

```python
from bisect import bisect_right
from itertools import accumulate


def f(x, positions, p_a, p_ax):
    k = bisect_right(positions, x)
    f_x = x * (2 * p_a[k] - p_a[-1]) + p_ax[-1] - 2 * p_ax[k]
    return f_x


def solution():
    N, Q = map(int, input().split())
    villages = [list(map(int, input().split())) for _ in range(N)]
    queries = [int(input()) for _ in range(Q)]
    villages.sort(key=lambda v: v[1])
    positions = [x for _, x in villages]
    p_a = [*accumulate([a for a, _ in villages], initial=0)]
    p_ax = [*accumulate([a * x for a, x in villages], initial=0)]
    answers = [0] * Q
    for i, q in enumerate(queries):
        f_q = f(q, positions, p_a, p_ax)
        answers[i] = f_q
    print("\n".join(map(str, answers)))


solution()

# input
# 3 1
# 2 1
# 1 3
# 3 6
# 4
```
