# 중간에서 만나기 (Meet In The Middle)

`-` 브루트 포스를 사용할 때 $N$개의 원소에 대해 탐색하지 않고 $\frac{N}{2}$개씩 쪼개서 탐색한 후 결합하여 시간 복잡도를 줄이는 알고리즘

`-` 결합하는 과정에서 $\alpha$ 만큼의 연산이 추가로 소요된다 ($\alpha$가 작아야 의미 있다)

`-` $O\left(2^N\right)$의 연산을 $O(\alpha)\cdot O\left(2^{^{\frac{N}{2}}}\right)$으로 줄이는 테크닉

## 카드 뽑기

- 문제 출처: [정올 5385번](https://jungol.co.kr/problem/5385)

`-` 카드를 절반씩 나누고 각 그룹마다 $1$장 이상 뽑아 만들 수 있는 정수와 해당 정수를 만드는 방법의 수를 딕셔너리에 저장하자. 그렇게 만든 각 딕셔너리에 $M$이 존재하면 등장 횟수만큼 카운팅하자

`-` 이제 하나의 딕셔너리를 순회할 것이다. 정수가 $x$이고 등장 횟수는 $c$라 하자. 반대편 딕셔너리에 $M-x$가 있고 등장 횟수가 $d$라면 $M$을 만드는 경우의 수는 $cd$이다. 해싱을 사용했으므로 $O(1)$에 계산 가능하다

`-` 최종적으로, 딕셔너리를 만드는 데 $O\left(N 2^{\frac{N}{2}}\right)$이고 합이 $M$인 경우의 수를 구하는 게 $O\left(2^{\frac{N}{2}}\right)$이므로 전체 알고리즘의 시간 복잡도는 $O\left(N 2^{\frac{N}{2}}\right)$이다

```python
from collections import defaultdict
from itertools import combinations


def make_all_subset_sums(s):
    n = len(s)
    sums = defaultdict(int)
    for k in range(1, n + 1):
        for e in combinations(s, k):
            sums[sum(e)] += 1
    return sums


def solution():
    N, M = map(int, input().split())
    array = list(map(int, input().split()))
    half = N // 2 
    left = array[:half]
    right = array[half:]
    left_sums = make_all_subset_sums(left)
    right_sums = make_all_subset_sums(right)
    count = left_sums[M] + right_sums[M]
    for x, c in left_sums.items():
        count += c * right_sums[M - x]
    print(count)


solution()

# input
# 4 10
# -2 1 9 3
```

## 다트(Dart)

- 문제 출처: [정올 1827번](https://jungol.co.kr/problem/1827)

`-` 문제 설명이 조금 모호한데 같은 부분을 중복해서 맞히면 점수도 중복으로 받는다

`-` $1,2,3$발이라면 브루트 포스로 해결할 수 있다. 하지만 $4$발은 불가능이다

`-` $2$발씩 나누고 각각에 대해 $1$발, $2$발의 다트를 던졌을 때 가능한 점수 배열을 만들자. 그리고 정렬 후 이분 탐색을 하자. 그럼 $O\left(N^2 \log N\right)$에 정답을 구할 수 있다

`-` 이때 $2$발씩 나누는 건 대칭적이므로 하나만 구해도 된다

```python
from bisect import bisect_right


def solution():
    N, M = map(int, input().split())
    P = [int(input()) for _ in range(N)]
    half = [p for p in P if p <= M] + [p1 + p2 for p1 in P for p2 in P if p1 + p2 <= M]
    half.sort()
    max_scoer = 0
    for x in half:
        score = x
        i = bisect_right(half, M - x)
        if i > 0:
            score += half[i - 1]
        max_scoer = max(score, max_scoer)
    print(max_scoer)


solution()

# input
# 3 21
# 16
# 11
# 2
```
