# 깊이 우선 탐색 (Depth-First Search)

`-` 시작 노드에서 출발해 갈 수 있는 데까지 깊이 탐색한 뒤 막히면 이전 분기로 되돌아가서 다른 경로를 탐색하는 알고리즘

## 두 로봇

- 문제 출처: [정올 3230번](https://jungol.co.kr/problem/3230)

`-` 트리이므로 두 정점 사이의 경로는 하나로 유일하다. 따라서 두 정점 사이의 거리에서 경로 내 간선의 최대 가중치를 차감하면 정답이다

`-` 역추적을 통해 최대 가중치를 찾자

```python
def dfs(graph, weights, source):
    n = len(graph)
    inf = float("inf")
    stack = [source]
    distances = [inf] * n
    distances[source] = 0
    predecessors = [None] * n
    while stack:
        u = stack.pop()
        for v in graph[u]:
            if distances[v] < inf:
                continue
            stack.append(v)
            distances[v] = distances[u] + weights[u, v]
            predecessors[v] = u
    return distances, predecessors


def track(sink, predecessors, weights):
    w_max = 0
    node = sink
    while predecessors[node] is not None:
        prev = predecessors[node]
        w = weights[prev, node]
        w_max = max(w, w_max)
        node = predecessors[node]
    return w_max


def solution():
    N, source, sink = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    weights = {}
    for _ in range(N - 1):
        u, v, w = map(int, input().split())
        graph[u].append(v)
        graph[v].append(u)
        weights[u, v] = weights[v, u] = w
    distances, predecessors = dfs(graph, weights, source)
    w_max = track(sink, predecessors, weights)
    answer = distances[sink] - w_max
    print(answer)


solution()

# input
# 5 1 5
# 1 2 1
# 2 3 2
# 3 4 3
# 4 5 1
```

## 키 순서

- 문제 출처: [정올 2462번](https://jungol.co.kr/problem/2462)

`-` 삽질을 많이 했다. 위상 정렬로 풀릴 것 같았지만 그렇지 않아서 태그를 봤다

`-` 그래프 상에 학생 번호를 노드로 나타내고 $a$번 학생의 키가 $b$번 학생의 키보다 작은 걸 $a\to b$ 단방향 간선으로 나타내자

`-` 두 학생의 키를 비교한 데이터이므로 모순이 없으니 그래프 상에 사이클이 존재하지 않음에 주목하자

`-` 임의의 $v$번 학생의 키 순서가 결정론적일려면 나보다 작은 학생이 $a$명, 나보다 큰 학생이 $b$명일 때 $a+b = n - 1$를 만족해야 한다. 따라서 노드 $v$로 이동할 수 있는 노드 개수를 $a$, 노드 $v$에서 이동할 수 있는 노드 개수를 $b$라 할 때 $a + b + 1 = n$이면 $v$번 학생의 키 순서는 결정론적이다

`-` $v$번 학생의 키가 몇 번째인지 확인하는 건 DFS를 통해 $O(N + M)$에 수행할 수 있으며 가능한 $v$의 개수는 $N$이므로 전체 알고리즘의 시간 복잡도는 $O\left(N^2 + NM\right)$이다

```python
def dfs(graph, source, sink, visited, key):
    stack = [source]
    while stack:
        u = stack.pop()
        for v in graph[u]:
            if visited[v] == key:
                continue
            stack.append(v)
            visited[v] = key
            if v == sink:
                return


def solution():
    N, M = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    for _ in range(M):
        a, b = map(int, input().split())
        graph[a].append(b)
    visited = [0] * (N + 1)
    key = 1
    answer = 0
    for v in range(1, N + 1):
        a = b = 0
        seen = set()
        for u in range(1, N + 1):
            if u == v:
                continue
            dfs(graph, u, v, visited, key)
            if visited[v] == key:
                a += 1
                seen.add(u)
            key += 1
        for t in range(1, N + 1):
            if t == v or t in seen:
                continue
            dfs(graph, v, t, visited, key)
            if visited[t] == key:
                b += 1
            key += 1
        if a + b + 1 == N:
            answer += 1
    print(answer)


solution()

# input
# 5 3
# 1 2
# 2 3
# 4 5
```
