# 깊이 우선 탐색 (Depth-First Search)

`-` 시작 노드에서 출발해 갈 수 있는 데까지 깊이 탐색한 뒤 막히면 이전 분기로 되돌아가서 다른 경로를 탐색하는 알고리즘

## 두 로봇

- 문제 출처: [정올 3230번](https://jungol.co.kr/problem/3230)

`-` 트리이므로 두 정점 사이의 경로는 하나로 유일하다. 따라서 두 정점 사이의 거리에서 경로 내 간선의 최대 가중치를 차감하면 정답이다

`-` 역추적을 통해 최대 가중치를 찾자

```python
def dfs(graph, weights, source):
    n = len(graph)
    inf = float("inf")
    stack = [source]
    distances = [inf] * n
    distances[source] = 0
    predecessors = [None] * n
    while stack:
        u = stack.pop()
        for v in graph[u]:
            if distances[v] < inf:
                continue
            stack.append(v)
            distances[v] = distances[u] + weights[u, v]
            predecessors[v] = u
    return distances, predecessors


def track(sink, predecessors, weights):
    w_max = 0
    node = sink
    while predecessors[node] is not None:
        prev = predecessors[node]
        w = weights[prev, node]
        w_max = max(w, w_max)
        node = predecessors[node]
    return w_max


def solution():
    N, source, sink = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    weights = {}
    for _ in range(N - 1):
        u, v, w = map(int, input().split())
        graph[u].append(v)
        graph[v].append(u)
        weights[u, v] = weights[v, u] = w
    distances, predecessors = dfs(graph, weights, source)
    w_max = track(sink, predecessors, weights)
    answer = distances[sink] - w_max
    print(answer)


solution()

# input
# 5 1 5
# 1 2 1
# 2 3 2
# 3 4 3
# 4 5 1
```
