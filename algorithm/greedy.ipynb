# 그리디 알고리즘 (Greedy)

`-` 매 순간 최적의 선택을 하여 전체 문제의 최적해를 구하는 알고리즘. 매 순간 최적인 것만 고려하면 되기에 동적 계획법보다 빠르게 동작한다. 단, 문제가 **탐욕스런 선택 속성**과 **최적 부분 구조**를 만족해야 올바르게 동작한다

`-` **탐욕스런 선택 속성**: 매 순간 최적의 선택이 이후의 선택에 영향을 주지 않는다

`-` **최적 부분 구조**: 전체 문제의 최적해가 부분 문제의 최적해로 구성된다

## 저울

- 문제 출처: [정올 2499번](https://jungol.co.kr/problem/2499)

`-` 단계별 문제를 풀어보자. 그리디 알고리즘을 사용하는 문제라고 한다

`-` 추를 사용하여 측정할 수 없는 최소 무게를 구해야 한다. 최솟값을 구하기 위해 무게를 기준으로 오름차순 정렬하자

`-` $i - 1$번째 추까지 이용해 $0$부터 $x$까지의 무게를 만들 수 있다고 해보자. $i$번째 추의 무게를 $w$라고 하자

`-` 만약 $w > x + 1$라면 $x+1$이 정답이다. 그렇지 않다면 $x$를 $x+w$로 갱신하면 된다. 최종적으로 순회를 끝마쳤다면 $x+1$이 정답이다. $x$의 초깃값은 $0$으로 설정하면 된다

`-` 전체 알고리즘의 시간 복잡도는 $O(N \log N)$이다

```python
def solution():
    N = int(input())
    weights = list(map(int, input().split()))
    weights.sort()
    x = 0
    for w in weights:
        if w > x + 1:
            break
        x += w
    print(x + 1)


solution()

# input
# 3
# 1 2 4
```

## 책꽂이 만들기

- 문제 출처: [정올 1929번](https://jungol.co.kr/problem/1929)

`-` 힌트 보고 풀었다. [모두더하기](https://jungol.co.kr/problem/1190) 문제를 풀면 도움이 된다고 한다. 이 문제는 단순히 매 선택마다 최솟값 $2$개를 뽑으면 풀린다

`-` 이 문제에선 길이가 $\sum\limits_{i=1}^{N} L_i$인 널빤지를 $N-1$번 잘라서 필요한 길이의 널빤지 $N$개로 만들어야 한다. 어떤식으로 잘라야 최적인지 가늠하기 어렵다

`-` 대신 거꾸로 생각하자. $N$개의 널빤지를 적절한 방식으로 이어 붙여 길이가 $\sum\limits_{i=1}^{N} L_i$인 널빤지 하나로 만드는 것이다. 그리고 이는 [모두더하기](https://jungol.co.kr/problem/1190) 문제와 동일하게 해결할 수 있다

`-` 발상의 전환이 필요한 문제였다

```python
import heapq


def solution():
    N = int(input())
    array = [int(input()) for _ in range(N)]
    heapq.heapify(array)
    cost = 0
    while len(array) >= 2:
        x = heapq.heappop(array)
        y = heapq.heappop(array)
        cost += x + y
        heapq.heappush(array, x + y)
    print(cost)


solution()

# input
# 4
# 1
# 2
# 3
# 4
```

## 허수아비

- 문제 출처: [정올 8566번](https://jungol.co.kr/problem/8566)

`-` $f(i)$는 $A_1$부터 $A_i$까지 고려했을 때 합이 $P$ 이상인 부분 집합 중 크기가 가장 작은 부분집합의 크기이다. 당연하게도 방어력이 큰 것을 사용하는 게 이득이다

`-` 최소 힙을 사용해서 여태까지 설치한 허수아비의 방어력을 관리하자. 매 $i$마다 $A_i$를 추가한 뒤 방어력 합이 $P$ 이상인 동안 힙에서 원소를 제거할 것이다. 결과적으로 힙의 크기가 $f(i)$가 된다

```python
import heapq


def solution():
    N, P = map(int, input().split())
    array = list(map(int, input().split()))
    fs = [-1] * N
    pq = []
    defense = 0
    for i, a in enumerate(array):
        heapq.heappush(pq, a)
        defense += a
        while pq and defense >= P + pq[0]:
            defense -= heapq.heappop(pq)
        if defense >= P:
            fs[i] = len(pq)
    print(*fs)      


solution()

# input
# 5 10
# 3 6 1 1 10
```

## 택배

- 문제 출처: [정올 2641번](https://jungol.co.kr/problem/2641)

`-` 각 마을에 방문하면 일단 물건 배송을 마친 뒤 가능한 최대로 박스를 싣자. 이때 받는 마을번호가 작은 걸 우선시 해야 한다. 그리고 이미 실은 박스더라도 마을번호를 더 줄일 수 있다면 그렇게 하자 (단, 개수 손해를 보면 안 된다)

`-` 이를 위해 마을마다 몇 개의 박스를 배송해야 되는지 기록하자. $i$번째 마을을 방문하면 기록해둔 배열을 바탕으로 해당 개수만큼 배달한 뒤 트럭 용량에서 차감하자. 그리고 $i$번째 마을에서 보내는 물건을 모두 실을 것이다. 그 후 트럭 용량을 초과하지 않는 선에서 마을번호가 큰 것을 제외하자. 이는 우선순위 큐를 이용하면 된다. 분할 배송이 가능한 점을 유의하자

`-` 전체 알고리즘의 시간 복잡도는 $O(N + M\log M)$이다

```python
import heapq


def solution():
    N, C = map(int, input().split())
    M = int(input())
    from_to = [[] for _ in range(N + 1)]
    for _ in range(M):
        s, e, c = map(int, input().split())
        from_to[s].append((e, c))
    pq = []
    delivery_counts = [0] * (N + 1)
    capacity = 0
    max_count = 0
    for i in range(1, N + 1):
        c = delivery_counts[i]
        capacity -= c
        max_count += c
        for e, c in from_to[i]:
            heapq.heappush(pq, (-e, c))
            delivery_counts[e] += c
            capacity += c
        while capacity > C:
            e, c = heapq.heappop(pq)
            e *= -1
            if e <= i:
                continue
            capacity -= c
            delivery_counts[e] -= c
            if capacity < C:
                heapq.heappush(pq, (-e, C - capacity))
                delivery_counts[e] += C - capacity
                capacity = C
    print(max_count)    


solution()

# input
# 2 10
# 1
# 1 2 20
```

## 냉장고

- 문제 출처: [정올 2641번](https://jungol.co.kr/problem/2641)

`-` $y$를 기준으로 오름차순 정렬하자. 모든 화학 물질을 냉장고에 보관해야 한다. $i-1$번 화학 물질까진 모두 냉장고에 보관했고 이제 $i$번 화학 물질을 냉장고에 보관할 차례라고 해보자

`-` $i+1$번 화학 물질을 $i$번 화학 물질과 같은 냉장고에 보관하고 싶다. 만약 보관 온도 구간이 겹치지 않는다면 둘은 같은 냉장고에 보관될 수 없다. 그리고 $y$를 기준으로 정렬했으므로 이후의 화학 물질도 마찬가지이다. 따라서 새로운 냉장고를 사용해야 한다

`-` 그렇지 않고 보관 온도 구간이 겹친다면 이들을 온도가 $y_i$인 냉장고 하나에 보관할 수 있다 

`-` 정렬은 $O(N \log N)$이며 각 화학 물질을 $1$번만 확인하면 되므로 총 $O(N)$이다. 따라서 전체 알고리즘의 시간 복잡도는 $O(N \log N)$이다

```python
def solution():
    N = int(input())
    array = [list(map(int, input().split())) for _ in range(N)]
    array.sort(key=lambda x: x[1])
    count = 1
    t = array[0][1]
    for x, y in array:
        if x <= t <= y:
            continue
        t = y
        count += 1
    print(count)


solution()

# input
# 4
# -15 5
# -10 36
# 10 73
# 27 44
```

## 비밀번호

- 문제 출처: [정올 2468번](https://jungol.co.kr/problem/2468)

`-` 먼저 $A$보다 작은 수 중 켜진 비트 수가 같고 가장 가까운 수를 찾자. $A$의 최하위 세팅된 비트를 $0$으로 만들고 우측 비트를 $1$로 바꾸면 된다

`-` 이것이 불가능하다면 최하위 세팅된 비트만 $0$으로 바꾸고 재귀적으로 수행하자. 이때 $1$로 바꿔야 하는 우측 비트의 개수를 하나 증가시켜야 한다. 만약 최하위 세팅된 비트가 없다면 정답은 $0$이다

`-` 이제 $A$보다 큰 수 중 켜진 비트 수가 같고 가장 가까운 수를 찾자. $A$의 최하위 세팅된 비트를 $0$으로 만들고 좌측 비트를 $1$로 바꾸면 된다

`-` 불가능하다면 최하위 세팅된 비트만 $0$으로 만들고 재귀적으로 작업을 반복하자. 이때 최하위 세팅된 비트를 $0$으로 바꾸고 좌측 비트를 $1$로 바꾼 뒤 최하위 비트부터 $k$개를 $1$로 바꾸면 된다. 초기 $k=0$이며 재귀를 수행할 때마다 $1$씩 증가시키자

`-` 참고로 큰 수 찾기를 구현할 땐 맨 앞에 비트 $0$을 추가하자. 기존 비트보다 더 많은 비트가 필요할 수 있기 때문이다

```python
def prev_permutation(binary, k):
    n = len(binary)
    i = lsb(binary)
    if i is None:
        return ["0"]
    binary[i] = "0"
    for j in range(1, k + 1):
        if not (i + j < n and binary[i + j] == "0"):
            return prev_permutation(binary, k + 1)
    binary[i + 1 : i + k + 1] = ["1"] * k
    return binary


def next_permutation(binary, k):
    n = len(binary)
    i = lsb(binary)
    binary[i] = "0"
    if i >= 1 and binary[i - 1] == "0":
        binary[i - 1] = "1"
        binary[n - k : n] = ["1"] * k
        return binary
    return next_permutation(binary, k + 1)


def lsb(binary):
    n = len(binary)
    for i in range(n - 1, -1, -1):
        if binary[i] == "1":
            return i


def solution():
    A = int(input())
    binary = bin(A)[2:]
    low = prev_permutation(list(binary), 1)
    high = next_permutation(["0"] + list(binary), 0)
    low = int("".join(low), 2)
    high = int("".join(high), 2)
    print(low, high)


solution()

# input
# 43
```

`-` 지문이 간접적이서 그렇지 이전 순열과 이후 순열을 구하는 문제와 같다고 하더라 (C++은 딸깍에 풀림)

## ABBC

- 문제 출처: [정올 5123번](https://jungol.co.kr/problem/5123)

`-` 기타 알고리즘으로 푸는 문제 (별 의미는 없다). 처음엔 [비드맨](https://www.acmicpc.net/problem/19590) 같은 문제인줄 알았다 (기타 알고리즘이라길래 더더욱)

`-` $A$는 뒤에 $B$가 있을 때만 지울 수 있다. $C$는 앞에 $B$가 있을 때만 지울 수 있다. 맨 뒤에 $A$부터 역순으로 순회하며 조건을 충족하는 $B$ 중 가장 뒤에 있는 것을 제거하자. 해당 $A$가 $B$와 매칭되면 나머지 $A$도 그 $B$와 매칭되므로 그리디하게 제거하자. $C$도 비슷한 논리로 처리하면 된다. 맨 앞에 $C$부터 순회하며 조건을 충족하는 $B$ 중 가장 앞에 있는 $B$를 제거하자

`-` $B$의 위치는 Deque으로 관리하면 편하다

```python
from collections import deque


def solution():
    S = input().rstrip()
    a_indices = [i for i, s in enumerate(S) if s == "A"]
    b_indices = deque([i for i, s in enumerate(S) if s == "B"])
    c_indices = [i for i, s in enumerate(S) if s == "C"]
    answer = 0
    for i_a in reversed(a_indices):
        if b_indices and i_a < b_indices[-1]:
            answer += 1
            b_indices.pop()
    for i_c in c_indices:
        if b_indices and i_c > b_indices[0]:
            answer += 1
            b_indices.popleft()
    print(answer)


solution()

# input
# ABCBBACBABB
```

`-` 비슷한 그리디 논리를 사용하는 문제로 [공항](https://www.acmicpc.net/problem/10775) 문제가 있다
