# 두 포인터 (Two Pointers)

`-` 두 개의 포인터를 사용해 구간이 변함에 따라 구간의 함숫값이 변하는 걸 효율적으로 갱신하는 탐색 기법

## 회전 초밥(고)

- 문제 출처: [정올 2577번](https://jungol.co.kr/problem/2577)

`-` 실질적으로 원형 배열을 표현하기 위해 앞의 $k$개의 원소를 뒤에 추가하자. 그럼 가능한 모든 경우를 고려할 수 있다

`-` 이제 슬라이딩 윈도우로 해결하면 되며 구간 내 초밥 종류별 개수, 유니크한 초밥 개수를 고려하자. 쿠폰 번호에 해당하는 초밥은 항상 먹을 수 있다고 생각하면 편하다

`-` 전체 알고리즘의 시간 복잡도는 $O(N + d)$이다

```python
def solution():
    N, d, k, c = map(int, input().split())
    array = [int(input()) for _ in range(N)]
    array.extend(array[:k])
    counts = [0] * (d + 1)
    counts[c] = float("inf")
    n_unique = 1
    for i in range(k):
        a_i = array[i]
        counts[a_i] += 1
        n_unique += counts[a_i] == 1
    max_count = n_unique
    for j in range(k, N + k):
        i = j - k
        a_i, a_j = array[i], array[j]
        counts[a_i] -= 1
        n_unique -= counts[a_i] == 0
        counts[a_j] += 1
        n_unique += counts[a_j] == 1
        max_count = max(n_unique, max_count)
    print(max_count)


solution()

# input
# 3 10 2 10
# 1
# 2
# 3
```
